## Encoding Categorical Data (Ordinal & Label)
Categorical data encoding is the process of converting text-based categories into numerical values so that machine learning algorithms can process them. Because most ML models rely on mathematical equations (like calculating Euclidean distances or matrix multiplications), they cannot interpret raw text strings directly.  

### 1.  Ordinal Encoding (Best for Ordinal Data)
This converts ordered categories into integers (0, 1, 2) while explicitly preserving their natural rank and hierarchy.When to use: Data where the sequence matters (e.g., sizes like S, M, L, XL).  

### 2. One-Hot Encoding (Best for Low-Cardinality Nominal Data)
This technique creates a new binary column (0 or 1) for each unique category in the feature. It ensures the model does not assume an incorrect numerical order between independent categories.When to use: Nominal data with few unique values (e.g., under 15 categories).  

### 3. Target Encoding / Mean Encoding (Best for High-Cardinality Data)
Replaces each category value with the expected mean of the target variable for that specific category.When to use: High-cardinality nominal features (e.g., ZIP codes, City names, Product IDs) where One-Hot Encoding would create thousands of columns.

## ONE HOT ENCODING
One-hot encoding is a data preprocessing technique that converts categorical text data into a binary numeric format (0s and 1s). It creates a new column for each unique category, placing a 1 where the category is present and a 0 everywhere else, preventing machine learning models from assuming false mathematical ranks.

The **dummy variable trap** is a statistical scenario where independent variables are perfectly correlated, creating a situation called perfect multicollinearity. This happens when you include a dummy variable for every category in a regression model alongside an intercept column, which breaks the underlying math of the model.  
For K input columns, we make k-1 columns deleting 1 column because where all columns are 0, then value must be of the deleted column.

```python
df=pd.read_csv('covid_toy.csv')
```

```python
from sklearn.model_selection import train_test_split as tts
X_train,X_test,y_train,y_test=tts(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

### -- Normal Zindgi

```python
# adding simple imputer to fever col
si=SimpleImputer()
X_train_fever=si.fit_transform(X_train[['fever']])
X_test_fever=si.fit_transform(X_test[['fever']])
X_train_fever.shape #null replaced by mean
```

```python
# ordinal encoding --> cough
oe=OrdinalEncoder(categories=[['Mild', 'Strong']])
X_train_cough=oe.fit_transform(X_train[['cough']])
X_test_cough=oe.fit_transform(X_test[['cough']])
X_train_cough.shape
```

```python
# ordinal encoding --> gender, city
ohe=OneHotEncoder(drop='first', sparse_output=False)
X_train_gender_city=ohe.fit_transform(X_train[['gender', 'city']])
X_test_gender_city=ohe.fit_transform(X_test[['gender', 'city']])
X_train_gender_city.shape
```

```python
#Extracting age
X_train_age=X_train.drop(columns=['gender', 'fever', 'cough', 'city']).values
X_test_age=X_test.drop(columns=['gender', 'fever', 'cough', 'city']).values
X_train_age.shape
```

```python

X_train_transformed = np.concatenate((X_train_age,X_train_fever,X_train_gender_city,X_train_cough),axis=1)
# also the test data
X_test_transformed = np.concatenate((X_test_age,X_test_fever,X_test_gender_city,X_test_cough),axis=1)

X_train_transformed
```
