## Tensors
A tensor is a mathematical container used to store numbers. Think of it as a grid or a box that holds data in a clean, organized layout.In machine learning, computers only understand numbers, so every piece of data (like an image, text, or audio) must be converted into a tensor before the system can read it.

**Types**
1. 0D
2. 1D
3. 2D
4. 3D

### 0D Tensor (Scalar)  
A single, isolated number with no rows, columns, or depth.  
Dimensions: 0 axes.  
Real-world ML example: A person's exact age, the price of one stock, or a single temperature reading.  
`tensor_0d = np.array(24)`

### 1D Tensor (Vector)  
A single row (or column) of numbers lined up in a row. You only need one index number to look up a value.  
Dimensions: 1 axis.  
Real-world ML example: A patient's vital signs package (e.g., Heart Rate, Blood Pressure, Body Temperature).  
`tensor_1d = np.array([72, 120, 98])`  

### 2D Tensor (Matrix)  
A flat grid or table containing rows and columns. You need two coordinates—a row number and a column number—to find a specific entry.  
Dimensions: 2 axes.  
Real-world ML example: A standard Excel spreadsheet tracking multiple items. For example, columns hold different house features (size, bedrooms, bathrooms) and each row represents a separate house.  
`tensor_2d = np.array([,  # House 1 features  
    [2400, 4, 3]   # House 2 features  
])`

### 3D Tensor (Cube)  
A stack of multiple 2D grids layered on top of each other. It forms a 3D block or cube of numbers. You need three coordinates (which page, which row, which column) to find a number.  
How to visualize it: A multi-page Excel workbook, or a physical Rubik's cube.  
Real-world ML example: Digital text data. If you have a sequence of sentences, Page 1 is Sentence 1, Page 2 is Sentence 2. Rows represent the words, and columns represent the math codes for those words.  
`[    
  [[1, 2], [3, 4]],  # Page 1 (Front slice)  
  [[5, 6], [7, 8]]   # Page 2 (Back slice)  
]`

### 4D Tensor (rank 4)
A 4D tensor is an array of 3D tensors. It has 4 axes or dimensions.In computer vision, a single color (RGB) image is stored as a 3D tensor with dimensions for height, width, and color channels. When you train a deep learning model, you do not feed images one by one; instead, you feed them in batches to optimize GPU performance
If you have a batch of \(32\) color images, where each image has a resolution of \(224 \times 224\) pixels, the data structure is represented as a 4D tensor with a shape of \((32, 224, 224, 3)\).
Axis 0 (Batch Size): Number of images in the current batch.  
Axis 1 (Height): The vertical pixel count of each image.  
Axis 2 (Width): The horizontal pixel count of each image.  
Axis 3 (Channels): The color channels (typically 3 for Red, Green, Blue).

### 5D Tensor (Rank 5)
A 5D tensor is an array of 4D tensors. It has 5 axes or dimensions.
Video data introduces a temporal dimension (time) because a video is a continuous sequence of individual image frames over time. A single video is a 4D tensor consisting of frames, height, width, and color channels. To process multiple videos concurrently during model training, a 5th dimension is required.
Axis 0 (Batch Size(Samples): Number of video clips in the batch.  
Axis 1 (Frames): The temporal sequence of image slices making up each video clip.  
Axis 2 (Height): The vertical pixel count of each frame.  
Axis 3 (Width): The horizontal pixel count of each frame.  
Axis 4 (Channels): The color channels (typically 3 for RGB)
